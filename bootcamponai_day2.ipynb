```python
import pandas as pd

data = pd.DataFrame({
    "Name": ["Aman123", "Rahul@", "Priya456", "Neha#"],
    "City": ["Delhi", "Mumbai", "Pune", "Agra"]
})

data["Name"] = data["Name"].str.replace(r"[^a-zA-Z]", "", regex=True)

print(data)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.title("Sine Wave")
plt.show()
```
