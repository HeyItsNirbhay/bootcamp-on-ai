```python
import random

numbers = [random.randint(1, 100) for i in range(10)]

print("Random numbers:", numbers)

print("Even numbers:")
for num in numbers:
    if num % 2 == 0:
        print(num)
```

```python
string = "Python programming is easy"

print("Original string:", string)

print("String in reverse:", string[::-1])

words = string.split()
print("Words in reverse:", " ".join(words[::-1]))
```
